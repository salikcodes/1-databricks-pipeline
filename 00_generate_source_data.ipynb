```python
import pandas as pd, random
from faker import Faker
fake = Faker(); random.seed(1)
base = "/Volumes/workspace/landing/raw"
def save(df, source, table, batch="batch1"):
    p = f"{base}/{source}/{table}/{batch}"
    dbutils.fs.mkdirs(p)
    df.to_csv(f"{p}/{table}.csv", index=False)

N = 5000
# MySQL
cust = pd.DataFrame({
  "cust_id": range(1, N+1),
  "name": [fake.name() for _ in range(N)],
  "email": [fake.email() for _ in range(N)],
  "city": [fake.city() for _ in range(N)],
  "last_updated_ts": pd.Timestamp("2026-01-01")})
save(cust, "mysql", "customers")
orders = pd.DataFrame({
  "order_id": range(1, 20001),
  "cust_id": [random.randint(1, N) for _ in range(20000)],
  "order_amt": [round(random.uniform(10, 900), 2) for _ in range(20000)],
  "order_dt": [fake.date_between("-90d", "today") for _ in range(20000)],
  "last_updated_ts": pd.Timestamp("2026-01-01")})
save(orders, "mysql", "orders")
# Teradata
save(pd.DataFrame({
  "txn_id": range(1, 20001),
  "acct_id": [random.randint(1, 3000) for _ in range(20000)],
  "txn_amt": [round(random.uniform(-500, 5000), 2) for _ in range(20000)],
  "txn_dt": [fake.date_between("-90d", "today") for _ in range(20000)]}),
  "teradata", "transactions")
# SAP HANA
save(pd.DataFrame({
  "gl_id": range(1, 10001),
  "gl_account": [random.choice(["4000","5000","6100","7200"]) for _ in range(10000)],
  "amount": [round(random.uniform(100, 9000), 2) for _ in range(10000)],
  "posting_dt": [fake.date_between("-90d", "today") for _ in range(10000)]}),
  "saphana", "gl_postings")
# Salesforce
save(pd.DataFrame({
  "opp_id": range(1, 3001),
  "account_name": [fake.company() for _ in range(3000)],
  "stage": [random.choice(["Prospect","Won","Lost"]) for _ in range(3000)],
  "amount": [round(random.uniform(1000, 90000), 2) for _ in range(3000)]}),
  "salesforce", "opportunities")
# POS
save(pd.DataFrame({
  "sale_id": range(1, 30001),
  "store_id": [random.randint(1, 50) for _ in range(30000)],
  "product": [random.choice(["Card","Loan","Insurance","Wallet"]) for _ in range(30000)],
  "qty": [random.randint(1, 5) for _ in range(30000)],
  "price": [round(random.uniform(5, 300), 2) for _ in range(30000)],
  "sale_ts": [fake.date_time_between("-30d", "now") for _ in range(30000)]}),
  "pos", "pos_sales")/Volumes/workspace/landing/raw/mysql/customers/batch1/customers.csv
```
